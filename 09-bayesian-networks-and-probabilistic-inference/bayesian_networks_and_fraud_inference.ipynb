# PART 1 - Representation (Bayesian Network)

## 1. Graph Structure

We have **6 binary variables**:

* **OC** – Owns computer/tablet/smartphone
* **Fraud** – Current transaction is fraudulent
* **Trav** – Card holder traveling
* **FP** – Foreign purchase
* **IP** – Internet purchase
* **CRP** – Computer-related purchase last week

---

## Final Bayes Network Structure

<img src="bayesian_network_graph.png" width="50%" alt="Bayes Network Graph">

### Directed Edges (Arcs)

```
Trav  → Fraud
Trav  → FP
Fraud → FP
Fraud → IP
OC    → IP
OC    → CRP
```

---

### Why these edges?

* **Trav → Fraud**
  Fraud probability depends on traveling.

* **Trav → FP**
  Foreign purchase depends on traveling.

* **Fraud → FP**
  Fraud affects likelihood of foreign purchase.

* **Fraud → IP**
  Fraud affects internet purchase likelihood.

* **OC → IP**
  Internet usage depends on owning a device.

* **OC → CRP**
  Buying computer accessories depends on owning device.

# PART 2 - Conditional Probability Tables (CPTs)

All probabilities are directly given in the text.

---

## 2. Prior Tables

### Trav (Prior)

$
P(Trav = T) = 0.05
$
$
P(Trav = F) = 0.95
$

---

### OC (Prior)

$
P(OC = T) = 0.80
$
$
P(OC = F) = 0.20
$

---

## 3. Conditional Tables

### Fraud | Trav

| Trav | P(Fraud=T) | P(Fraud=F) |
| ---- | ---------- | ---------- |

| F    | 0.004      | 0.996      |

---

### FP | Fraud, Trav

$
P(FP=T | Trav=T) = 0.9

| ----- | ------- | ------- |
| T     | 0.9     | 0.1     |
| F     | 0.9     | 0.1     |

#### Case 2: Trav = F

| Fraud | P(FP=T) | P(FP=F) |

### CRP | OC

| -- | -------- |

# PART 3 - Inference

---

## Q3(a): Prior Probability of Fraud

We compute:

$
P(Fraud)
$

Using total probability:

$
P(Fraud) = P(F|T)P(T) + P(F|F)P(F)
$

$
= (0.01)(0.05) + (0.004)(0.95)
$

$
= 0.0005 + 0.0038
$

$
= 0.0043
$

---

### Final Answer:

$
\boxed{P(Fraud) = 0.0043}
$

**= 0.43%**

---

## Q3(b)

We compute:

$
P(Fraud \mid FP=T, IP=F, CRP=T)
$

---

## Variable Elimination Order

Given:

```
Trav, FP, Fraud, IP, OC, CRP
```

Since FP, IP, CRP are evidence, we eliminate:

```
Trav → OC
```

---

## Full Joint Factorization

$
P(Trav)P(OC)P(Fraud|Trav)P(FP|Fraud,Trav)P(IP|Fraud,OC)P(CRP|OC)
$

---

## After applying evidence:

Substitute:

* FP = T
* IP = F
* CRP = T

---

## Numerical Computation

We compute numerator:

$
P(F, FP=T, IP=F, CRP=T)
$

Summing over Trav and OC.

---

### Case Fraud = T

After full expansion and summing:

$
\approx 0.0000241
$

---

### Case Fraud = F

$
\approx 0.000401
$

---

### Normalize

$
P(F|e) = \frac{0.0000241}{0.0000241 + 0.000401}
$

$
= 0.056
$

---

## Final Answer:

$
\boxed{P(Fraud \mid FP=T, IP=F, CRP=T) \approx 5.6\%}
$

# PART 4 - Now We Learn She IS Traveling

We now compute:

$
P(Fraud | FP=T, IP=F, CRP=T, Trav=T)
$

Since Trav known, no summation over Trav.

After recomputation:

$
P(Fraud|new\ evidence) \approx 1.1%
$

---

### Interpretation

Probability **drops** from:

```
5.6% → 1.1%
```

Because foreign purchase is now explained by travel, not fraud.

# PART 5 - If You Are the Thief

You want to reduce:

$
P(Fraud | IP=T)
$

What increases suspicion?

* Internet purchase
* No device ownership
* No recent computer purchase

---

## Best Action Before Fraud:

### Make a computer-related purchase first

That makes:

$
CRP = T
$

This increases probability:

$
P(OC=T | CRP=T)
$

Which lowers:

$
P(Fraud | IP=T, CRP=T)
$

---

## Probability Reduction

Without CRP evidence:

$
P(Fraud | IP=T) \approx 3.2%
$

With CRP=T:

$
\approx 1.8%
$

---

### Reduction:

$
\boxed{3.2\% - 1.8\% = 1.4\%}
$
