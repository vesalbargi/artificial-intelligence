### Question 1: Iterated Elimination of Strictly Dominated Strategies and Best Response Method

Consider a two-player game where **Player 1** chooses among strategies **U**, **M**, or **D**, and **Player 2** chooses among strategies **L**, **C**, or **R**.

The payoff matrix is given below, where each cell lists
**(Player 1’s payoff, Player 2’s payoff)**.

|       | **L**  | **C**  | **R**  |
| ----- | ------ | ------ | ------ |
| **U** | (3, 0) | (3, 1) | (3, 2) |
| **M** | (1, 3) | (4, 4) | (1, 3) |
| **D** | (2, 1) | (2, 2) | (2, 0) |

**a)** Apply the **iterated elimination of strictly dominated strategies (IESDS)** to simplify the game. Show each step clearly and explain which strategies are eliminated and why.

**Step 1: Eliminate dominated strategies for Player 1**

Compare **U** and **D**:

* Against **L**: U = 3, D = 2 → U better
* Against **C**: U = 3, D = 2 → U better
* Against **R**: U = 3, D = 2 → U better

**U strictly dominates D**

**Eliminate strategy D**

**Step 2: Reduced game**

|       | **L** | **C** | **R** |
| ----- | ----- | ----- | ----- |
| **U** | (3,0) | (3,1) | (3,2) |
| **M** | (1,3) | (4,4) | (1,3) |

**Step 3: Eliminate dominated strategies for Player 2**

Compare **L** and **R**:

* Against **U**: L = 0, R = 2 → R better
* Against **M**: L = 3, R = 3 → equal

Since **R is never worse and sometimes better**, **R strictly dominates L**.

**Eliminate strategy L**

**Step 4: Final reduced game**

|       | **C** | **R** |
| ----- | ----- | ----- |
| **U** | (3,1) | (3,2) |
| **M** | (4,4) | (1,3) |

No further strictly dominated strategies exist.

**b)** Using the **best response method** on the remaining game after IESDS, identify **all pure-strategy Nash equilibria**.

**Player 1 best responses**

* If Player 2 plays **C** → Player 1 prefers **M** (4 > 3)
* If Player 2 plays **R** → Player 1 prefers **U** (3 > 1)

**Player 2 best responses**

* If Player 1 plays **U** → Player 2 prefers **R** (2 > 1)
* If Player 1 plays **M** → Player 2 prefers **C** (4 > 3)

**Pure Strategy Nash Equilibria**

1. **(M, C)**
2. **(U, R)**

Both outcomes are **mutual best responses**.

### Question 3: Mixed Strategy Nash Equilibrium, Strict Dominance in Mixed Strategies, and Calculating Payoffs

Consider a two-player game where **Player 1** chooses among strategies **A, B,** or **C**, and **Player 2** chooses between strategies **X** or **Y**.
The payoff matrix below shows **(Player 1’s payoff, Player 2’s payoff)**.

|       | X      | Y      |
| ----- | ------ | ------ |
| **A** | (1, 3) | (5, 1) |
| **B** | (5, 1) | (1, 3) |
| **C** | (2, 4) | (2, 0) |

**a)** Show that strategy **C** is strictly dominated by a **mixed strategy** for Player 1.
Specify the mixed strategy (a combination of **A** and **B**) that strictly dominates **C**, and verify this by comparing payoffs against both **X** and **Y**.

Let Player 1 mix between **A** and **B**:

* Play **A** with probability **½**
* Play **B** with probability **½**

Expected payoff vs **X**

$
0.5(1) + 0.5(5) = 3 > 2
$

Expected payoff vs **Y**

$
0.5(5) + 0.5(1) = 3 > 2
$

Mixed strategy strictly dominates **C**

**Eliminate C**

**b)** Find the **mixed strategy Nash equilibrium**, assuming Player 1 does not play **C** due to strict dominance.
Let Player 1 play **A** with probability ( p ) and **B** with probability ( 1 - p ), and let Player 2 play **X** with probability ( q ) and **Y** with probability ( 1 - q ).

Let:

* Player 1: A with probability **p**, B with **1 − p**
* Player 2: X with probability **q**, Y with **1 − q**

**Player 1 indifference condition**

$
\text{Payoff(A)} = \text{Payoff(B)}
$

$
1q + 5(1-q) = 5q + 1(1-q)
$

$
5 - 4q = 1 + 4q
$

$
q = 0.5
$

**Player 2 indifference condition**

$
3p + 1(1-p) = 1p + 3(1-p)
$

$
2p + 1 = 3 - 2p
$

$
p = 0.5
$

**Mixed Strategy Nash Equilibrium**

* Player 1: A (0.5), B (0.5)
* Player 2: X (0.5), Y (0.5)

**c)** Calculate the **expected payoffs** for both players in this mixed strategy Nash equilibrium.

**Player 1**

$
E = 0.5(3) + 0.5(3) = 3
$

**Player 2**

$
E = 0.5(2) + 0.5(2) = 2
$
